```python
#Q1. Top 10 IPL teams by total wins — Bar Chart
#level 1
# Team-wise total wins
team_wins = df['winner'].value_counts().head(10)

# Chart
plt.figure(figsize=(12, 6))

plt.bar(
    team_wins.index,
    team_wins.values,
    color='skyblue',
    edgecolor='black'
)

plt.title('Top 10 IPL Teams by Total Wins', fontsize=16)
plt.xlabel('Teams', fontsize=12)
plt.ylabel('Total Wins', fontsize=12)

plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
#Q2. Har season mein kitne matches hue — Line Chart
#level 1
# Season-wise match count
season_matches = df.groupby('season').size()

plt.figure(figsize=(12, 6))

plt.plot(
    season_matches.index,
    season_matches.values,
    marker='o',
    linestyle='-',
    linewidth=2,
    color='green'
)

plt.title('IPL Matches Played in Each Season', fontsize=16)
plt.xlabel('Season', fontsize=12)
plt.ylabel('Total Matches', fontsize=12)

plt.xticks(season_matches.index)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
#Q9. Har season mein teams ki wins compare karo — Grouped Bar Chart
#level 2
# Season and team-wise wins
season_team_wins = (
    df.groupby(['season', 'winner'])
      .size()
      .reset_index(name='total_wins')
)

# Top 5 teams overall
top_5_teams = df['winner'].value_counts().head(5).index

# Pivot table
pivot_wins = season_team_wins[
    season_team_wins['winner'].isin(top_5_teams)
].pivot(
    index='season',
    columns='winner',
    values='total_wins'
).fillna(0)

# Grouped bar chart
ax = pivot_wins.plot(
    kind='bar',
    figsize=(15, 7),
    width=0.8
)

plt.title('Season-wise Wins: Top 5 IPL Teams', fontsize=16)
plt.xlabel('Season')
plt.ylabel('Total Wins')

plt.xticks(rotation=0)
plt.legend(title='Teams', bbox_to_anchor=(1.02, 1))
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
#Q11. Har season mein toss decision bat vs field — Stacked Bar Chart
#level 2
# Season-wise toss decision count
toss_data = pd.crosstab(
    df['season'],
    df['toss_decision']
)

# Stacked bar chart
toss_data.plot(
    kind='bar',
    stacked=True,
    figsize=(13, 6),
    color=['orange', 'skyblue'],
    edgecolor='black'
)

plt.title('Season-wise Toss Decisions', fontsize=16)
plt.xlabel('Season')
plt.ylabel('Total Toss Decisions')

plt.xticks(rotation=0)
plt.legend(title='Toss Decision')

plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
#level 2
#Q13. Runs se jeetne wale matches ka margin distribution — Histogram
# Runs se jeetne wale matches
run_margin = df.loc[
    df['result'] == 'runs',
    'result_margin'
].dropna()

plt.figure(figsize=(12, 6))

plt.hist(
    run_margin,
    bins=15,
    color='orange',
    edgecolor='black',
    alpha=0.8
)

plt.title('Distribution of Winning Margins (Runs)', fontsize=16)
plt.xlabel('Winning Margin in Runs')
plt.ylabel('Number of Matches')

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
#Q17. Season aur team-wise wins ka Heatmap
#level 3
# Season and team-wise wins
season_team_wins = (
    df.groupby(['season', 'winner'])
      .size()
      .reset_index(name='total_wins')
)

# Heatmap ke liye pivot table
heatmap_data = season_team_wins.pivot(
    index='winner',
    columns='season',
    values='total_wins'
).fillna(0)

# Heatmap
plt.figure(figsize=(16, 9))

plt.imshow(
    heatmap_data,
    aspect='auto',
    cmap='YlGnBu'
)

plt.colorbar(label='Total Wins')

plt.title('IPL Team Wins Across Seasons', fontsize=16)
plt.xlabel('Season')
plt.ylabel('Teams')

plt.xticks(
    range(len(heatmap_data.columns)),
    heatmap_data.columns
)

plt.yticks(
    range(len(heatmap_data.index)),
    heatmap_data.index
)

plt.tight_layout()
plt.show()
```

```python
#Q19. Top 5 teams ki season-wise performance — Multi-Line Chart
#level 3
# Top 5 teams overall
top_5_teams = df['winner'].value_counts().head(5).index

# Season-wise team wins
season_team_wins = (
    df.groupby(['season', 'winner'])
      .size()
      .reset_index(name='total_wins')
)

# Top 5 teams filter
top_team_data = season_team_wins[
    season_team_wins['winner'].isin(top_5_teams)
]

# Pivot table
pivot_data = top_team_data.pivot(
    index='season',
    columns='winner',
    values='total_wins'
).fillna(0)

# Multi-line chart
plt.figure(figsize=(14, 7))

for team in pivot_data.columns:
    plt.plot(
        pivot_data.index,
        pivot_data[team],
        marker='o',
        linewidth=2,
        label=team
    )

plt.title('Season-wise Performance of Top 5 IPL Teams', fontsize=16)
plt.xlabel('Season')
plt.ylabel('Total Wins')

plt.xticks(pivot_data.index)
plt.legend(title='Teams', bbox_to_anchor=(1.02, 1))
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Q23. Har team ke matches played vs matches won — Grouped Bar Chart
#level 3
# Matches played
matches_played = pd.concat(
    [df['team1'], df['team2']]
).value_counts()

# Matches won
matches_won = df['winner'].value_counts()

# DataFrame create
team_stats = pd.DataFrame({
    'Matches Played': matches_played,
    'Matches Won': matches_won
}).fillna(0)

# Top 10 teams by matches played
team_stats = team_stats.sort_values(
    'Matches Played',
    ascending=False
).head(10)

# Grouped bar chart
team_stats.plot(
    kind='bar',
    figsize=(14, 7),
    width=0.8
)

plt.title('Matches Played vs Matches Won by IPL Teams', fontsize=16)
plt.xlabel('Teams')
plt.ylabel('Number of Matches')

plt.xticks(rotation=45, ha='right')
plt.legend(title='Statistics')

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
